1. PyPREP
Library digunakan khusus untuk standardisasi dan otomatisasi preprocessing data EEG (elektroensefalografi/rekaman aktivitas otak). PyPREP merupakan implementasi Python dari PREP Pipeline yang populer di MATLAB.
Fungsi Utama: Mendeteksi dan menangani saluran (channel) sinyal EEG yang rusak atau buruk secara otomatis.Kegunaan: Melakukan interpolasi saluran yang rusak dan menghitung referensi rata-rata yang kuat (robust average reference) agar data EEG bersih sebelum masuk ke tahap analisis atau machine learning.
2. Scipy adalah library dasar yang sangat populer untuk komputasi ilmiah, matematika, dan teknik di Python. SciPy dibangun di atas NumPy dan menyediakan banyak modul efisien.Fungsi Utama: Menyediakan algoritma siap pakai untuk manipulasi data tingkat lanjut.Kegunaan:Pemrosesan Sinyal: Memfilter noise pada sinyal (seperti ECG atau EEG) menggunakan modul scipy.signal.Optimasi: Mencari nilai minimum/maksimum dari suatu fungsi matematika.Statistik & Integrasi: Melakukan uji statistik, transformasi Fourier, dan menyelesaikan persamaan diferensial.
3. wandb adalah sebuah platform dan library MLOps yang digunakan untuk melacak, memvisualisasikan, dan mengelola eksperimen machine learning (ML/DL).Fungsi Utama: Bertindak sebagai "kotak hitam" atau jurnal digital otomatis untuk model AI Anda.Kegunaan:Eksperimen Tracking: Mencatat loss, akurasi, dan metrik model secara real-time dalam bentuk grafik interaktif.Hyperparameter Tuning: Membantu mencari kombinasi pengaturan model terbaik (menggunakan fitur W&B Sweeps).Versi Model & Data: Menyimpan riwayat arsitektur model dan dataset yang digunakan agar eksperimen mudah diulang kembali.
4. pyECG merupakan sebuah library khusus yang dirancang untuk analisis dan pemrosesan sinyal ECG (elektrokardiogram/rekaman aktivitas jantung).Fungsi Utama: Mengekstrak informasi penting dari grafik detak jantung.Kegunaan:Deteksi Gelombang: Menemukan puncak gelombang R (titik tertinggi detak jantung) dan kompleks QRS secara otomatis.Analisis HRV: Menghitung Heart Rate Variability (variabilitas detak jantung) untuk analisis kesehatan fisik atau stres.Ekstraksi Fitur: Mengubah sinyal mentah jantung menjadi data numerik yang siap digunakan untuk diagnosis medis atau input model AI.

2. Dampak Energi dan Lingkungan akibat AIDi balik kecerdasannya, infrastruktur AI membutuhkan pusat data (data center) berskala masif yang sangat haus energi dan air.

* Krisis Air Bersih (Water Footprint): Server AI menghasilkan panas yang luar biasa. Untuk mendinginkannya, pusat data menyedot jutaan liter air bersih. Pemasangan pusat data di wilayah yang rawan kekeringan dapat mengancam pasokan air bagi masyarakat lokal.
* Sampah Elektronik (E-waste): Siklus hidup perangkat keras AI (seperti GPU khusus) sangat pendek karena teknologi terus berkembang cepat. Perserikatan Bangsa-Bangsa (UN) memproyeksikan infrastruktur AI akan menghasilkan hingga 2,5 juta ton sampah elektronik per tahun per 2030.

3. Cara Mengatasi Dampak Lingkungan (Pendapat Pribadi & Solusi). Menurut pandangan saya, penyelesaian masalah lingkungan akibat AI tidak bisa hanya dibebankan pada konsumen dengan mengurangi pemakaian, melainkan membutuhkan revolusi struktural dari hulu ke hilir oleh korporasi dan pemerintah. Berikut beberapa langkah strategis yang harus diambil:
* Kewajiban Penggunaan Energi Terbarukan (Hulu): Pemerintah harus mengetatkan regulasi sehingga perusahaan teknologi wajib membangun atau mendanai pembangkit listrik ramah lingkungan (seperti tenaga surya, angin, atau nuklir generasi baru) langsung di dekat pusat data mereka, bukan sekadar membeli "sertifikat karbon" di atas kertas.
* Inovasi Arsitektur Algoritma yang Efisien (Tengah): Komunitas pengembang AI harus menggeser fokus mereka. Dibanding terus-menerus memperbesar ukuran model (yang memakan komputasi besar), fokus harus diarahkan pada Green AI—yaitu merancang algoritma yang lebih kecil, lebih padat, namun memiliki performa setara (contohnya teknik distillation dan quantization), sehingga penggunaan energinya saat dijalankan jauh lebih hemat.
* Sistem Pendingin Ramah Lingkungan (Hilir): Pusat data wajib beralih dari pendingin berbasis air evaporatif ke sistem sirkulasi tertutup (closed-loop cooling) atau pendingin cair (liquid immersion cooling) yang tidak menghabiskan air bersih di wilayah sekitarnya.
